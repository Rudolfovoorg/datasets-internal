```python
import json
import glob
import pandas as pd
import os
```

```python
def flatten_dict(d, prefix='', level=0):
    for i,j in d.items():
        if isinstance(j, dict):
            yield from flatten_dict(j, f'{prefix}{i}_', level+1)
        else:
            yield f'{prefix}{i}', j

def data_reader(filelist):
    for filename in filelist:
        with open(filename) as f:
            yield dict(flatten_dict(json.load(f)))
```

```python
mask_no_estimate = 'solutions/without_estimate/*/*.output.json'
mask_with_estimate = 'solutions/with_estimate/*/*.output.json'
mask_inputs = 'with_sol/*/*.json'
```

```python
df_no_estimate = pd.DataFrame(data_reader(glob.glob(mask_no_estimate)))
df_no_estimate['instance'] = df_no_estimate['meta_data_instance'].apply(lambda x: x.replace('.json', '').split('/')[-1])
df_no_estimate['instance']
```

```python
df_with_estimate = pd.DataFrame(data_reader(glob.glob(mask_with_estimate)))
df_with_estimate['instance'] = df_with_estimate['meta_data_instance'].apply(lambda x: x.replace('.json', '').split('/')[-1])
df_with_estimate['instance']
```

```python
df_input_data = pd.DataFrame(data_reader(glob.glob(mask_inputs)))
df_input_data.rename(columns={
    'info_instance': 'instance'
}, inplace=True)
df_input_data
```

```python
df_filtered_input_data = df_input_data[['instance', 'offset', 'initial_estimate',
                                        'info_additional_info_initial_astimate_algo', 'info_additional_info_time_limit']].copy()
# df_input_data
df_filtered_input_data['size'] = df_input_data['qubo_shape'].apply(
    lambda x: x[0])
df_filtered_input_data.rename(columns={
    'info_additional_info_initial_astimate_algo': 'initial_estimate_heuristic',
    'info_additional_info_time_limit': 'initial_estimate_time_limit'
}, inplace=True)
df_filtered_input_data.set_index('instance', inplace=True)
df_filtered_input_data
```

```python
selected_columns = [
    'instance',
    'qubo_computed_val',
    'meta_data_time',
    'meta_data_root_node_time',
    'meta_data_root_node_heuristic_run_count',
    'meta_data_root_node_sdp_value',
    'meta_data_root_node_heuristic_value',
    'meta_data_root_node_total_heur_time',
    'meta_data_root_node_root_solution'
]
```

```python
df_filtered_with_estimate = df_with_estimate[selected_columns].copy()

df_filtered_with_estimate.rename(columns={
    'qubo_computed_val': 'final_qubo_value_w',
    'meta_data_time': 'total_time_w',
    'meta_data_root_node_time': 'root_eval_time_w',
    'meta_data_root_node_heuristic_run_count': 'heur_run_count_w',
    'meta_data_root_node_sdp_value': 'sdp_value_w',
    'meta_data_root_node_heuristic_value': 'heur_value_w',
    'meta_data_root_node_total_heur_time': 'heur_total_time_w',
    'meta_data_root_node_root_solution': 'root_solution_w'
}, inplace=True)

df_filtered_with_estimate.set_index('instance', inplace=True)

df_filtered_no_estimate = df_no_estimate[selected_columns].copy()

df_filtered_no_estimate.rename(columns={
    'qubo_computed_val': 'final_qubo_value',
    'meta_data_time': 'total_time',
    'meta_data_root_node_time': 'root_eval_time',
    'meta_data_root_node_heuristic_run_count': 'heur_run_count',
    'meta_data_root_node_sdp_value': 'sdp_value',
    'meta_data_root_node_heuristic_value': 'heur_value',
    'meta_data_root_node_total_heur_time': 'heur_total_time',
    'meta_data_root_node_root_solution': 'root_solution'
}, inplace=True)
df_filtered_no_estimate.set_index('instance', inplace=True)
```

```python
df_merged = pd.concat([df_filtered_input_data, df_filtered_with_estimate, df_filtered_no_estimate], axis=1)
df_merged['heur_value_with-no'] = df_merged['heur_value_w'] - df_merged['heur_value']
df_merged['sdp_value_with-no'] = df_merged['sdp_value_w'] - df_merged['sdp_value']
df_merged['gap'] = df_merged['sdp_value'] - df_merged['heur_value']
df_merged['gap_w'] = df_merged['sdp_value_w'] - df_merged['heur_value_w']
df_merged['gap_w-no'] = df_merged['gap_w'] - df_merged['gap']
```
