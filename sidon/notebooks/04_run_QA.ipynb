```python
import glob
import os
import json

from sidion_qubo import load_data, get_G_ising, solve_ising, export_results_to_json, Ising
from dwave.system import DWaveSampler, FixedEmbeddingComposite
```

```python
solver_type = 'Advantage2_system1.8'
case_path = '../data/{}/send_{}/m{}/case{}/beta{}/'
#list_of_cases = sorted(glob.glob(case_path.format(solver_type, '*', 1, 0, 0.0)))
list_of_cases = sorted(glob.glob(case_path.format(solver_type, '*', '*', '*', 0.0)))

isings = ['J_log', 'J_qac']
num_reads = [100]

#assert len(list_of_cases) == 1 # for testing
```

```python
for case in list_of_cases:
    for ising_type in isings:
        with open(f'{case}{ising_type}.json') as f:
            data = json.load(f)

        ising = Ising().deserialize(data)

        for n in num_reads:
            print(f'Solving {ising} with num_reads={n} ...')
            emb = {node: [node] for node in ising.nodes}
            sampler = FixedEmbeddingComposite(DWaveSampler(solver=solver_type), embedding=emb)

            result = solve_ising(
                ising,
                sampler,
                num_reads=n,
                annealing_time=1000,
                auto_scale=True,
                reduce_intersample_correlation=True)
            print(' ... Done.')

            
            output_file_folder = case.replace(f'data/{solver_type}', 'results/QA')
            os.makedirs(os.path.dirname(output_file_folder), exist_ok=True)
            output_file_name = f"{output_file_folder}QA_{ising_type}_num_reads_{n}.json"
            print(f'Exporting results to {output_file_name} ...')
            export_results_to_json(result, output_file_name)
            print(' ... Done.')
```
