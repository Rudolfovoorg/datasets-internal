```python
import glob
import os
from neal import SimulatedAnnealingSampler
import json

from sidion_qubo import load_data, get_G_ising, solve_ising, export_results_to_json, Ising
```

```python
solver_type = 'Advantage2_system1.8'
case_path = '../data/{}/send_{}/m{}/case{}/beta{}/'
list_of_cases = sorted(glob.glob(case_path.format(solver_type, '*', '*', '*', 0.0)))
isings = ['J_log', 'J_qac']
num_reads = [10, 100, 1000]

#assert len(list_of_cases) == 1 # for testing
```

```python
list_of_cases
```

```python
for case in list_of_cases:
    for ising_type in isings:
        with open(f'{case}{ising_type}.json') as f:
            data = json.load(f)

        ising = Ising().deserialize(data)

        for n in num_reads:
            print(f'Solving {ising} with num_reads={n} ...')
            result = solve_ising(ising, SimulatedAnnealingSampler(), num_reads=n)
            print(' ... Done.')

            
            output_file_folder = case.replace(f'data/{solver_type}', 'results/SA')
            os.makedirs(os.path.dirname(output_file_folder), exist_ok=True)
            output_file_name = f"{output_file_folder}SA_{ising_type}_num_reads_{n}.json"
            print(f'Exporting results to {output_file_name} ...')
            export_results_to_json(result, output_file_name)
            print(' ... Done.')

```
